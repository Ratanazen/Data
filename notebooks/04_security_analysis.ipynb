# 🛡️ LogShield — Notebook 04: Security Threat Detection & Risk Scoring

This notebook demonstrates the LogShield threat detection engine: reading security incidents from Gold Parquet, inspecting risk scores (0–100), examining forensic indicators, and generating automated WAF mitigation commands.

```python
import pandas as pd
import pyarrow.parquet as pq

sec_file = "data/gold/gold_security_events.parquet"
df_sec = pq.read_table(sec_file).to_pandas()
print(f"Total Flagged Security Incidents: {len(df_sec):,}")
print(df_sec.head(5)[['event_id', 'timestamp', 'ip', 'event_type', 'severity', 'risk_score']])
```

## 1. Distribution of Incident Severity Levels

```python
print(df_sec['severity'].value_counts())
print("\nAverage Risk Score:", df_sec['risk_score'].mean())
print("Maximum Risk Score:", df_sec['risk_score'].max())
```

## 2. Generate Automated WAF & iptables Drop Commands

```python
high_risk_ips = df_sec[df_sec['risk_score'] >= 50]['ip'].unique()
print(f"Flagged {len(high_risk_ips)} High/Critical Threat IPs for Perimeter Isolation:\n")
for ip in high_risk_ips[:10]:
    print(f"iptables -A INPUT -s {ip} -j DROP # LogShield WAF Drop")
```
