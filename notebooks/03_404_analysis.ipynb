# 📊 LogShield — Notebook 03: 404 Error Analytics & Peak Time Modeling

This notebook explores the core analytical findings of Project #19 / LogShield: identifying the 404 error volume (43,153 errors, 21.58%), confirming the peak failure window (03:00–04:00 AM UTC with 4,655 errors), and plotting the hourly distribution and day-of-week intensity heatmap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read precomputed hourly dataset
df_hourly = pd.read_csv("hourly_404_errors.csv")
print(df_hourly.head())
print("Total 404 errors:", df_hourly["error_count"].sum())
```

## 1. Plot 24-Hour 404 Error Distribution

```python
plt.figure(figsize=(12, 5))
colors = ['#ff0055' if h == 3 else '#0284c7' for h in df_hourly['hour']]
plt.bar(df_hourly['hour'], df_hourly['error_count'], color=colors, edgecolor='black', alpha=0.9)
plt.title("404 Error Occurrences by Hour of Day (Peak: 03:00 AM)", fontsize=14, fontweight='bold')
plt.xlabel("Hour of Day (UTC)", fontsize=11)
plt.ylabel("404 Error Count", fontsize=11)
plt.xticks(range(24))
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
